```python
def fix_outliers_iqr(df):
    num=df.select_dtypes(include='number').columns
    for col in num:
        Q1=df[col].quantile(0.25)
        Q3=df[col].quantile(0.75)
        IQR= Q3 - Q1
        lower_bound=  Q1 - 1.5 * IQR
        upper_bound= Q3 + 1.5 * IQR
        df[col]=df[col].apply(lambda x: lower_bound if x < lower_bound else upper_bound if x > upper_bound else x)
    return df 
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

cols = ['job', 'marital', 'default', 'housing', 'loan',
        'contact', 'month', 'day_of_week']

one = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

encoded = one.fit_transform(df[cols])

cols_encoded = one.get_feature_names_out(cols)

df_encoded = pd.DataFrame(encoded, columns=cols_encoded, index=df.index)

df_encoded = pd.concat([df.drop(columns=cols), df_encoded], axis=1)

df_encoded
```

```python
df_encoded['poutcome'] = df_encoded['poutcome'].map({
    'nonexistent': 0,
    'failure': 1,
    'success': 2
})

df_encoded.head()
```

```python
df1=df_encoded
```

```python
from imblearn.over_sampling import SMOTE

# Separate features and target
X = df1.drop('y', axis=1)
y = df1['y']

# Apply SMOTE
smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X, y)

print("Before SMOTE:", y.value_counts())
print("After SMOTE:", y_smote.value_counts())
```

```python
skb = SelectKBest(score_func=f_classif,k=25)
```

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
X_scaled = ss.fit_transform(X_selected)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y ,test_size=0.2,random_state=40)
```

```python
print('X_train Shape:',X_train.shape)
print('X_test Shape:',X_test.shape)
print('y_train Shape:',y_train.shape)
print('y_test Shape:',y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
LR = LogisticRegression()
```

```python
LR.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score
y_pred_LR = LR.predict(X_test)
accuracy = accuracy_score(y_test,y_pred_LR)
print("Accuracy:",accuracy)
```

```python
from sklearn.metrics import classification_report, f1_score,precision_score, recall_score
```

```python
accuracy_LR= accuracy_score(y_test,y_pred_LR)
precision_LR= precision_score(y_test,y_pred_LR)
f1_LR= f1_score(y_test,y_pred_LR)
recall_LR= recall_score(y_test,y_pred_LR)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
```

```python
# Selecting Algorithm 
models = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(),
    "Naive bayes": GaussianNB(),
}                                                   
```

```python
#Store results
scores = []
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy= accuracy_score(y_test,y_pred)
    precision= precision_score(y_test,y_pred)
    f1= f1_score(y_test,y_pred)
    recall= recall_score(y_test,y_pred)
    scores.append({
        "Model": name,
        "accuracy": round(accuracy, 2),
        "precision": round(precision, 2),
        "f1": round(f1, 2),
        "recall": round(recall, 2)
    })
# Convert to DataFrame for nice display
results_df = pd.DataFrame(scores)
print(results_df)
```
